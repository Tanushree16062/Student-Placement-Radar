```python
from pathlib import Path

journal_text = """
July 2026 ✨

This month has been a crazy walkthrough for me.

Back in June, I was completely empty-handed.
Worried about what I would be doing ahead.
Placements, internships, studies, career,
personal life tensions...
everything was sitting on my head together.

But then came July...

I couldn't have imagined what was waiting for me.

A ton of opportunities came by,
and this time I didn't leave a single one.

I talked to so many people.
I told my stories.
I heard theirs.

And most importantly...

I stopped saying NO.

Because there comes a phase in life
where you simply need to grab onto time,
keep moving,
and trust that destiny will someday
knock on your door.

And I am not kidding.

Things begin changing.

What you need to do is stop saying:

"I can't do it."

Instead say:

"Yes. Let me try this time."

Because sometimes opportunities are visitors.
If they leave,
they may never return the same way again.

So at last,
a shoutout to everyone out there
who is in chaos,
struggling,
sad,
or disappointed.

Listen to me carefully.

Step out from that corner.

The light is only a short step away.

You will find it.

Please keep going.

I know you are stronger than you think.

And I know...

you will never give up.

— July 2026 Journal
"""

html_content = f"""
<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">
<title>July 2026 Journal</title>

<style>
* {{
    margin:0;
    padding:0;
    box-sizing:border-box;
}}

body {{
    overflow:hidden;
    height:100vh;
    display:flex;
    justify-content:center;
    align-items:center;
    background:linear-gradient(-45deg,#0f172a,#1e293b,#3b2f2f,#111827);
    background-size:400% 400%;
    animation:bgmove 15s ease infinite;
    font-family:'Courier New', monospace;
}}

@keyframes bgmove {{
    0% {{background-position:0% 50%;}}
    50% {{background-position:100% 50%;}}
    100% {{background-position:0% 50%;}}
}}

.journal {{
    width:80%;
    max-width:900px;
    height:80vh;
    background:rgba(255,248,230,0.95);
    border-radius:24px;
    box-shadow:0 20px 60px rgba(0,0,0,.4);
    padding:40px;
    overflow:auto;
    position:relative;
}}

.title {{
    text-align:center;
    font-size:2rem;
    margin-bottom:20px;
    color:#5b4636;
}}

#text {{
    white-space:pre-wrap;
    line-height:1.9;
    color:#3d2e25;
    font-size:1.15rem;
}}

.cursor {{
    display:inline-block;
    width:10px;
    animation:blink .8s infinite;
}}

@keyframes blink {{
    50% {{opacity:0;}}
}}

.note {{
    position:absolute;
    font-size:24px;
    opacity:.15;
    animation:float 12s linear infinite;
}}

@keyframes float {{
    from {{
        transform:translateY(100vh);
    }}
    to {{
        transform:translateY(-120px);
    }}
}}
</style>
</head>

<body>

<div class="note" style="left:10%">🎷</div>
<div class="note" style="left:25%; animation-delay:2s;">☕</div>
<div class="note" style="left:40%; animation-delay:5s;">🎵</div>
<div class="note" style="left:60%; animation-delay:1s;">✨</div>
<div class="note" style="left:80%; animation-delay:3s;">📖</div>

<div class="journal">
    <div class="title">July 2026 • The Turnaround Month</div>
    <div id="text"></div><span class="cursor">|</span>
</div>

<script>
const text = `{journal_text}`;
const target = document.getElementById("text");

let i = 0;

function typeWriter() {{
    if(i < text.length) {{
        target.innerHTML += text.charAt(i);
        i++;
        setTimeout(typeWriter, 35);
    }}
}}

typeWriter();
</script>

</body>
</html>
"""

file_name = "July_2026_Cozy_Jazz_Journal.html"

Path(file_name).write_text(html_content, encoding="utf-8")

print(f"Saved: {{file_name}}")

# For Google Colab download
try:
    from google.colab import files
    files.download(file_name)
except:
    print("Open the HTML file in your browser.")
```
